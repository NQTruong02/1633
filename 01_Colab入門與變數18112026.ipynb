## 1-2 `print()` 與註解

- `print()` 會把括號裡的內容顯示出來，可以用逗號 `,` 一次印好幾個東西。
- `#` 後面的文字是**註解**，Python 不會執行，是寫給人看的說明。

```python
# 這一行是註解，不會被執行
print("今天的課程是", "Python 基礎")   # 逗號會自動加一個空格
print(3 + 5)                           # 也可以直接印計算結果
```

💡 **小技巧**：在 Colab 中，儲存格的**最後一行**如果是一個值，就算沒寫 `print()` 也會自動顯示。

```python
10 * 3
```

## 1-3 變數

變數就像一個「貼了名字的盒子」，用 `=` 把資料放進去。
注意：程式中的 `=` 是「**指定**」（把右邊的值放進左邊），不是數學上的「等於」。

```python
name = "小明"
age = 16
height = 170.5

print(name, age, height)
```

```python
# 變數的值可以被修改
age = 17
print("明年的年齡：", age)

# 也可以用變數本身來計算新的值
age = age + 1
print("再過一年：", age)
```

**變數命名規則**
- 可以用英文字母、數字、底線 `_`，但**不能用數字開頭**（`1score` ❌、`score1` ✅）
- **大小寫不同**：`Score` 和 `score` 是兩個不同的變數
- 不要用 Python 保留字（如 `print`、`if`、`for`、`list`）當變數名稱
- 建議取有意義的名字，例如 `math_score` 比 `a` 好懂

## 1-4 基本資料型態

| 型態 | 名稱 | 範例 |
|---|---|---|
| `int` | 整數 | `16`, `-3`, `0` |
| `float` | 浮點數（小數） | `170.5`, `3.14`, `2.0` |
| `str` | 字串（文字） | `"小明"`, `'A'`, `"123"` |
| `bool` | 布林值（真假） | `True`, `False` |

用 `type()` 可以查看資料的型態：

```python
print(type(16))
print(type(170.5))
print(type("小明"))
print(type(True))
```

⚠️ 加了引號就是**字串**，即使內容是數字也一樣。請比較下面兩格的結果：

```python
print(3 + 4)       # 數字相加
print("3" + "4")   # 字串相接
```

## 1-5 型態轉換

處理資料時常常需要轉換型態，例如從檔案讀進來的數字有時是字串。

| 函式 | 功能 |
|---|---|
| `int()` | 轉成整數（小數部分直接捨去） |
| `float()` | 轉成浮點數 |
| `str()` | 轉成字串 |

```python
score_text = "85"
score = int(score_text)
print(score + 10)

print(float("72.5"))
print(int(9.8))          # 小數直接捨去，不是四捨五入
print("分數是" + str(score))
```

## 1-7 常見錯誤訊息

看到紅色錯誤訊息不要緊張，最後一行通常會告訴你問題在哪：

| 錯誤 | 常見原因 |
|---|---|
| `NameError` | 變數名稱打錯、大小寫不同，或還沒執行定義它的那格 |
| `SyntaxError` | 語法錯誤，例如引號或括號沒有成對 |
| `TypeError` | 型態不合，例如 `"分數" + 85`（字串加數字） |
| `ValueError` | 值無法轉換，例如 `int("abc")` |
